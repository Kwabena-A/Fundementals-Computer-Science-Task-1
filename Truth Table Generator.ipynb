```python
variables = ['m', 'c', 'p', 't', 'o']
```

```python
dataframe = {}
table_length = 2 ** len(variables)
split_at = int(table_length/2)
for x in variables:
    print(split_at)
    new_col = [True for x in range(split_at)] + [False for x in range(split_at)]
    repeat_count = int(table_length / len(new_col))
    dataframe[x] = new_col * repeat_count
    split_at /= 2
    split_at = int(split_at)
```

```python
def truth_output(values: dict) -> bool:
    if ((values["c"] or values["t"]) and (not values["o"]) and (not values["p"])) or (values["m"]):
        return True
    return False
```

```python
df = pd.DataFrame(dataframe)
```
