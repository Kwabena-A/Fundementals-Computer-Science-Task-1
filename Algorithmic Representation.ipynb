```python
def door_closed(m, p, o, c, t) -> bool:
    if ((c or t) and (not o) and (not p)) or (m):
        return True
    return False
```

```python
x = door_closed(
    m = True, # True when Train is in motion
    p = False, # True when a passenger has requested that the door should open
    o = True, # True when objects are obstructing door closing
    c = True, # True when the conductor has requested that doors should close
    t = True # True when doors have been opened for 15s or more
)
x
```
